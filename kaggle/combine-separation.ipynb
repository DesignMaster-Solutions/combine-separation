# Does the Combine see separation?

Reproduction notebook for our NFL Big Data Bowl 2027 Open Track entry. It runs the analysis scripts from the public repo [DesignMaster-Solutions/combine-separation](https://github.com/DesignMaster-Solutions/combine-separation) on the competition data, in order, and prints each summary table. The chain it tests: a WR's 10 Hz Combine movement (break entry, speed kept through the cut) vs. the separation he creates on NFL routes in 2023-25 game tracking, then the same break detector on game cuts, split by man/zone and by defender distance.

```python
import glob, os, subprocess, sys
cands = [p for p in glob.glob('/kaggle/input/**/combine_tracking.csv', recursive=True)]
assert cands, 'competition data not attached'
DATA = os.path.dirname(cands[0]); print(DATA)
subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/DesignMaster-Solutions/combine-separation', '/kaggle/working/repo'], check=True)
```

```python
import pandas as pd
pd.set_option('display.width', 200)
def run(script, summary):
    r = subprocess.run([sys.executable, f'/kaggle/working/repo/analysis/{script}', DATA], capture_output=True, text=True)
    print(r.stdout[-3000:]); print(r.stderr[-3000:])
    r.check_returncode()
    if summary: display(pd.read_csv(f'/kaggle/working/repo/out/{summary}').head(20))
```

## 1. Combine route drills and 3-cone vs. separation (baseline)
